```python
# ==========================================
# 🚀 1️⃣ CHARGEMENT ET ANALYSE DU DATASET
# ==========================================

import os
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# 📂 Chemin du dataset
DATASET_PATH = "/content/dataset_yoga16/yoga16-dataset"

# 📏 Paramètres globaux
IMAGE_SIZE = (224, 224)  # Taille des images
BATCH_SIZE = 32          # Taille du batch
SEED = 42                # Reproductibilité

# 🧐 Vérification de la structure du dataset
def check_dataset(dataset_path):
    """Affiche la répartition des images par classe dans chaque dossier"""
    for subset in ["train", "test", "valid"]:
        subset_path = os.path.join(dataset_path, subset)
        print(f"\n📂 Vérification du dossier {subset.upper()} ({subset_path}):")

        if not os.path.exists(subset_path):
            print(f"⚠️ Dossier {subset} introuvable!")
            continue

        # Nombre d'images par classe
        class_counts = {c: len(os.listdir(os.path.join(subset_path, c)))
                        for c in sorted(os.listdir(subset_path))
                        if os.path.isdir(os.path.join(subset_path, c))}

        print(f"✅ {len(class_counts)} classes trouvées. {sum(class_counts.values())} images.")

check_dataset(DATASET_PATH)

# ==========================================
# 🎨 2️⃣ PRÉTRAITEMENT & AUGMENTATION DES IMAGES
# ==========================================

# 🚀 Augmentation des données pour améliorer la robustesse du modèle
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,   # Rotation légère pour simuler des variations
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,  # Appliqué uniquement aux poses symétriques
    brightness_range=[0.8, 1.2],
    fill_mode='nearest',
    validation_split=0.2   # 20% des données seront utilisées pour la validation
)

# 📂 Générateurs de données
train_gen = train_datagen.flow_from_directory(
    os.path.join(DATASET_PATH, "train"),
    target_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='training',
    seed=SEED
)

val_gen = train_datagen.flow_from_directory(
    os.path.join(DATASET_PATH, "train"),
    target_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='validation',
    seed=SEED
)

test_datagen = ImageDataGenerator(rescale=1./255)
test_gen = test_datagen.flow_from_directory(
    os.path.join(DATASET_PATH, "test"),
    target_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False
)

print(f"\n✅ Train: {train_gen.samples} images | Val: {val_gen.samples} | Test: {test_gen.samples}")

# ==========================================
# 🏗️ 3️⃣ CONSTRUCTION DU MODÈLE (EfficientNetB0)
# ==========================================

import tensorflow as tf
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras import layers, models, optimizers

def build_yoga_model(input_shape=(224, 224, 3), num_classes=16):
    """Construit un modèle basé sur EfficientNetB0 pour la classification des postures de yoga"""

    # 🔥 Chargement du modèle EfficientNet pré-entraîné
    base_model = EfficientNetB0(
        include_top=False,
        weights='imagenet',
        input_shape=input_shape
    )

    # 🔓 Décongélation des dernières couches pour le fine-tuning
    for layer in base_model.layers[-20:]:
        layer.trainable = True

    # 🏗️ Architecture du modèle
    model = models.Sequential([
        base_model,
        layers.GlobalAveragePooling2D(),
        layers.BatchNormalization(),
        layers.Dense(512, activation='relu'),
        layers.Dropout(0.5),  # ⚡ Évite l'overfitting
        layers.Dense(num_classes, activation='softmax')
    ])

    # 🎯 Compilation du modèle
    model.compile(
        optimizer=optimizers.Adam(learning_rate=1e-4),
        loss='categorical_crossentropy',
        metrics=['accuracy', tf.keras.metrics.Precision(), tf.keras.metrics.Recall()]
    )

    return model

yoga_model = build_yoga_model()
yoga_model.summary()

# ==========================================
# ⚖️ 4️⃣ AJOUT DE CLASS WEIGHTS (Gestion du déséquilibre)
# ==========================================

from sklearn.utils.class_weight import compute_class_weight

# 📊 Calcul des poids pour compenser les classes minoritaires
class_weights = compute_class_weight('balanced', classes=np.unique(train_gen.classes), y=train_gen.classes)
class_weights_dict = dict(enumerate(class_weights))

# ==========================================
# 🎯 5️⃣ ENTRAÎNEMENT DU MODÈLE
# ==========================================

from tensorflow.keras.callbacks import ReduceLROnPlateau, ModelCheckpoint, EarlyStopping

# 🚀 Callbacks pour améliorer l'entraînement
callbacks = [
    ModelCheckpoint('best_yoga_model.keras', monitor='val_accuracy', save_best_only=True, mode='max'),
    EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=5, min_lr=1e-6)  # 🔥 Réduction du LR si stagnation
]

# 🚀 Entraînement du modèle
history = yoga_model.fit(
    train_gen,
    epochs=50,
    validation_data=val_gen,
    callbacks=callbacks,
    class_weight=class_weights_dict
)

# ==========================================
# 🏆 6️⃣ ÉVALUATION DU MODÈLE
# ==========================================

# 🏅 Charger le meilleur modèle sauvegardé
best_model = models.load_model('best_yoga_model.keras')

# 🎯 Évaluer la performance sur le test set
test_loss, test_acc, test_prec, test_rec = best_model.evaluate(test_gen)
print(f"\n🔥 Performance finale :")
print(f"- Accuracy: {test_acc*100:.2f}%")
print(f"- Precision: {test_prec*100:.2f}%")
print(f"- Recall: {test_rec*100:.2f}%")

# ==========================================
# 📊 7️⃣ VISUALISATION DES RÉSULTATS
# ==========================================

from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns

# 📌 Prédictions du modèle
y_true = test_gen.classes
y_pred = np.argmax(best_model.predict(test_gen), axis=1)

# 🏗️ Matrice de confusion
plt.figure(figsize=(12, 10))
sns.heatmap(confusion_matrix(y_true, y_pred), annot=True, fmt='d')
plt.title('Matrice de Confusion')
plt.show()

# 📜 Rapport détaillé
print(classification_report(y_true, y_pred, target_names=list(train_gen.class_indices.keys())))
```

```python
from google.colab.output import eval_js
from base64 import b64decode
import numpy as np
import cv2

def capture_image():
    """Capture une image depuis la webcam dans Colab"""
    js_code = """
        async function captureImage() {
            const video = document.createElement('video');
            video.style.display = 'none';
            document.body.appendChild(video);

            const stream = await navigator.mediaDevices.getUserMedia({ video: true });
            video.srcObject = stream;
            await video.play();

            await new Promise(r => setTimeout(r, 1000)); // Attendre 1 sec pour stabiliser

            const canvas = document.createElement('canvas');
            canvas.width = video.videoWidth;
            canvas.height = video.videoHeight;
            canvas.getContext('2d').drawImage(video, 0, 0);
            const data = canvas.toDataURL('image/jpeg');

            stream.getTracks().forEach(track => track.stop());
            video.remove();
            return data;
        }
        captureImage();
    """
    image_data = eval_js(js_code)
    img_bytes = b64decode(image_data.split(",")[1])

    # Convertir en image OpenCV
    img_array = np.frombuffer(img_bytes, dtype=np.uint8)
    img = cv2.imdecode(img_array, cv2.IMREAD_COLOR)
    return img
```

```python
# ==========================================
# 🛠 8️⃣ DÉTECTION D'ERREURS DE POSTURE
# ==========================================

# Définir les points de repère MediaPipe
mp_pose = mp.solutions.pose

# Dictionnaire des erreurs et corrections par posture
POSE_ERRORS = {
    "Chaturanga": [
        {
            "name": "Épaules trop basses",
            "check": lambda l: (l[mp_pose.PoseLandmark.LEFT_SHOULDER.value].y > l[mp_pose.PoseLandmark.LEFT_ELBOW.value].y or
                               l[mp_pose.PoseLandmark.RIGHT_SHOULDER.value].y > l[mp_pose.PoseLandmark.RIGHT_ELBOW.value].y),
            "correction": "Maintenir les épaules alignées avec les coudes"
        },
        {
            "name": "Bassin trop haut",
            "check": lambda l: (l[mp_pose.PoseLandmark.LEFT_HIP.value].y < l[mp_pose.PoseLandmark.LEFT_SHOULDER.value].y - 0.1 or
                               l[mp_pose.PoseLandmark.RIGHT_HIP.value].y < l[mp_pose.PoseLandmark.RIGHT_SHOULDER.value].y - 0.1),
            "correction": "Engager les abdominaux pour abaisser le bassin"
        }
    ],
    "Downward_Facing_Dog": [
        {
            "name": "Dos arrondi",
            "check": lambda l: abs(l[mp_pose.PoseLandmark.LEFT_SHOULDER.value].y - l[mp_pose.PoseLandmark.LEFT_HIP.value].y) < 0.1,
            "correction": "Allonger la colonne vertébrale et pousser les hanches vers le haut"
        }
    ]
}

def detect_pose_errors(pose_name, landmarks):
    """Détecte les erreurs pour une posture donnée"""
    if pose_name not in POSE_ERRORS:
        return []

    errors = []
    for error in POSE_ERRORS[pose_name]:
        if error["check"](landmarks):
            errors.append(error)

    return errors

# Modifier la fonction process_frame
def process_frame(frame):
    """Détecte les keypoints et prédit la posture avec vérification d'erreurs"""
    results = pose_detector.process(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))

    if results.pose_landmarks:
        # Prédiction de la posture
        img = cv2.resize(frame, (224, 224))
        img = np.expand_dims(img/255.0, axis=0)
        prediction = model.predict(img)
        class_id = np.argmax(prediction)

        # Détection des erreurs
        landmarks = results.pose_landmarks.landmark
        return class_id, landmarks

    return -1, None

# ==========================================
# 📊 9️⃣ AFFICHAGE DES RÉSULTATS AMÉLIORÉ
# ==========================================

# Capturer une nouvelle image
frame = capture_image()

# Traiter l'image
class_id, landmarks = process_frame(frame)

# Afficher les résultats
plt.figure(figsize=(10, 8))
plt.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))

if class_id != -1:
    predicted_pose = index_to_class[class_id]
    errors = detect_pose_errors(predicted_pose, landmarks)

    title = f"Posture: {predicted_pose}"
    if errors:
        title += "\nErreurs détectées:"
        for i, error in enumerate(errors, 1):
            title += f"\n{i}. {error['name']} → {error['correction']}"
    else:
        title += "\n✅ Posture correcte!"

    plt.title(title, fontsize=12, pad=20)
else:
    plt.title("Aucune posture détectée")

plt.axis('off')
plt.show()
```

```python
# ==========================================
# 🛠 MISE À JOUR DU DICTIONNAIRE D'ERREURS
# ==========================================
import re
import inspect
POSE_ERRORS = {
    "Chaturanga": [
        {
            "name": "Épaules trop basses",
            "check": lambda l: (
                (l[mp_pose.PoseLandmark.LEFT_SHOULDER.value].y >
                 l[mp_pose.PoseLandmark.LEFT_ELBOW.value].y + 0.03) or
                (l[mp_pose.PoseLandmark.RIGHT_SHOULDER.value].y >
                 l[mp_pose.PoseLandmark.RIGHT_ELBOW.value].y + 0.03)
            ),
            "correction": "Maintenir les épaules au niveau des coudes\n→ Imaginez pousser le sol loin de vous"
        },
        {
            "name": "Bassin surélevé",
            "check": lambda l: (
                (l[mp_pose.PoseLandmark.LEFT_HIP.value].y <
                 l[mp_pose.PoseLandmark.LEFT_SHOULDER.value].y - 0.15) or
                (l[mp_pose.PoseLandmark.RIGHT_HIP.value].y <
                 l[mp_pose.PoseLandmark.RIGHT_SHOULDER.value].y - 0.15)
            ),
            "correction": "Engager les muscles abdominaux\n→ Ramener le nombril vers la colonne vertébrale"
        },
        {
            "name": "Alignement avant-bras",
            "check": lambda l: (
                abs(l[mp_pose.PoseLandmark.LEFT_WRIST.value].x -
                 l[mp_pose.PoseLandmark.LEFT_ELBOW.value].x) > 0.05 or
                abs(l[mp_pose.PoseLandmark.RIGHT_WRIST.value].x -
                 l[mp_pose.PoseLandmark.RIGHT_ELBOW.value].x) > 0.05
            ),
            "correction": "Alignez les poignets sous les coudes\n→ Former un angle de 90° avec les avant-bras"
        }
    ],
    # Ajouter d'autres postures au besoin
}

# ==========================================
# 🔍 FONCTION DE DÉTECTION AMÉLIORÉE
# ==========================================

def detect_pose_errors(pose_name, landmarks):
    """Version améliorée avec vérification de visibilité"""
    errors = []

    if pose_name not in POSE_ERRORS:
        return errors

    for error in POSE_ERRORS[pose_name]:
        try:
            # Vérifier la visibilité des points clés
            keypoints_visible = all(
                landmarks[point].visibility > 0.7
                for point in get_required_points(error)
            )

            if keypoints_visible and error["check"](landmarks):
                errors.append(error)
        except:
            continue

    return errors


def get_required_points(error):
    """Identifie les points clés nécessaires pour une vérification d'erreur"""
    lambda_code = inspect.getsource(error["check"]).replace("lambda l: ", "")

    # Détection des landmarks utilisés dans le code du lambda
    pattern = r"l\[mp_pose\.PoseLandmark\.([A-Z_]+)\.value\]"
    matches = re.findall(pattern, lambda_code)

    # Conversion des noms en valeurs d'enum
    required_landmarks = []
    for landmark_name in set(matches):  # set() pour éviter les doublons
        try:
            landmark_value = getattr(mp_pose.PoseLandmark, landmark_name).value
            required_landmarks.append(landmark_value)
        except AttributeError:
            continue

    return required_landmarks
```

```python
# ==========================================
# 📦 IMPORTS
# ==========================================

from google.colab import drive
import os
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
import cv2
import json
import mediapipe as mp
import re
import inspect
from tensorflow.keras.preprocessing.image import ImageDataGenerator, load_img, img_to_array
from google.colab import files

# ==========================================
# 🌐 CONFIGURATION MEDIAPIPE
# ==========================================

mp_pose = mp.solutions.pose
mp_drawing = mp.solutions.drawing_utils
pose_detector = mp_pose.Pose(
    min_detection_confidence=0.7,
    min_tracking_confidence=0.7,
    model_complexity=2
)

# ==========================================
# 🧠 CHARGEMENT DU MODÈLE
# ==========================================

drive.mount('/content/drive')
model = tf.keras.models.load_model("/content/drive/MyDrive/saved_yogaAI/best_yoga_model.keras")

# ==========================================
# 📚 MAPPAGE DES CLASSES
# ==========================================

with open("/content/drive/MyDrive/saved_yogaAI/classification_report.json", "r") as f:
    classification_data = json.load(f)

class_names = [key for key in classification_data.keys() if "avg" not in key and key != "accuracy"]
index_to_class = {i: class_names[i] for i in range(len(class_names))}

# ==========================================
# 🧮 FONCTIONS UTILITAIRES
# ==========================================

def calculate_angle(a, b, c):
    """Calcule l'angle entre trois points en degrés"""
    a = np.array(a)
    b = np.array(b)
    c = np.array(c)

    radians = np.arctan2(c[1]-b[1], c[0]-b[0]) - np.arctan2(a[1]-b[1], a[0]-b[0])
    angle = np.abs(radians*180.0/np.pi)

    return angle if angle <= 180 else 360 - angle

def get_required_points(error):
    """Identifie les points clés nécessaires pour une vérification"""
    try:
        lambda_code = inspect.getsource(error["check"]).replace("lambda l: ", "")
        pattern = r"l\[mp_pose\.PoseLandmark\.([A-Z_]+)\.value\]"
        matches = re.findall(pattern, lambda_code)

        return list(set([
            getattr(mp_pose.PoseLandmark, name).value
            for name in matches
            if hasattr(mp_pose.PoseLandmark, name)
        ]))
    except Exception as e:
        print(f"Erreur dans get_required_points: {e}")
        return []

# ==========================================
# 🛠 ERREURS DE POSTURE
# ==========================================

POSE_ERRORS = {
    "Chaturanga": [
        {
            "name": "Épaules trop basses",
            "check": lambda l: (
                l[mp_pose.PoseLandmark.LEFT_SHOULDER.value].y >
                l[mp_pose.PoseLandmark.LEFT_ELBOW.value].y + 0.03
            ),
            "correction": "Alignez les épaules avec les coudes\n→ Poussez le sol loin de vous"
        }
    ],
    "Seated_Forward_Bend_Pose": [
        {
            "name": "Dos arrondi",
            "check": lambda l: calculate_angle(
                [l[mp_pose.PoseLandmark.LEFT_HIP.value].x, l[mp_pose.PoseLandmark.LEFT_HIP.value].y],
                [l[mp_pose.PoseLandmark.LEFT_SHOULDER.value].x, l[mp_pose.PoseLandmark.LEFT_SHOULDER.value].y],
                [l[mp_pose.PoseLandmark.LEFT_EAR.value].x, l[mp_pose.PoseLandmark.LEFT_EAR.value].y]
            ) < 150,
            "correction": "Allongez la colonne vertébrale vers l'avant"
        }
    ]
}

# ==========================================
# 🔍 VÉRIFICATIONS GÉNÉRIQUES
# ==========================================

def check_generic_alignment(landmarks):
    """Vérifications communes à toutes les postures"""
    errors = []

    # Alignement vertical
    try:
        left_shoulder = landmarks[mp_pose.PoseLandmark.LEFT_SHOULDER.value]
        left_hip = landmarks[mp_pose.PoseLandmark.LEFT_HIP.value]

        if abs(left_shoulder.x - left_hip.x) > 0.2:
            errors.append({
                "name": "Alignement latéral incorrect",
                "correction": "Maintenez l'alignement épaule-hanche"
            })
    except:
        pass

    return errors

# ==========================================
# 📸 FONCTION D'ANALYSE PRINCIPALE
# ==========================================

def analyze_yoga_pose():
    """Analyse complète d'une posture de yoga"""
    uploaded = files.upload()
    if not uploaded:
        print("❌ Aucune image sélectionnée")
        return

    image_path = list(uploaded.keys())[0]
    img = cv2.imread(image_path)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Paramètres
    LOW_CONFIDENCE_THRESHOLD = 40.0
    warnings = []
    errors = []

    # Détection des points clés
    results = pose_detector.process(img_rgb)

    # Prédiction de la posture
    img_processed = cv2.resize(img, (224, 224)) / 255.0
    prediction = model.predict(np.expand_dims(img_processed, axis=0))
    class_id = np.argmax(prediction)
    confidence = np.max(prediction) * 100
    pose_name = index_to_class.get(class_id, "Inconnue")

    # Gestion des résultats
    if confidence < LOW_CONFIDENCE_THRESHOLD:
        warnings.append(f"⚠️ Prédiction peu fiable ({confidence:.1f}%)")

    if results.pose_landmarks:
        # Visualisation améliorée
        annotated_image = img_rgb.copy()
        mp_drawing.draw_landmarks(
            annotated_image,
            results.pose_landmarks,
            mp_pose.POSE_CONNECTIONS,
            landmark_drawing_spec=mp_drawing.DrawingSpec(color=(0,255,0), thickness=2),
            connection_drawing_spec=mp_drawing.DrawingSpec(color=(255,165,0), thickness=2)
        )

        # Détection des erreurs
        errors = []
        if pose_name in POSE_ERRORS:
            for error in POSE_ERRORS[pose_name]:
                try:
                    required_points = get_required_points(error)
                    if all(results.pose_landmarks.landmark[p].visibility > 0.6 for p in required_points):
                        if error["check"](results.pose_landmarks.landmark):
                            errors.append(error)
                except:
                    continue

        errors += check_generic_alignment(results.pose_landmarks.landmark)

        if pose_name not in POSE_ERRORS:
            warnings.append("ℹ️ Aucune vérification spécifique disponible")

    # Affichage des résultats
    plt.figure(figsize=(12, 8))
    plt.imshow(annotated_image if results.pose_landmarks else img_rgb)

    title = f"POSTURE : {pose_name.upper()}\nConfiance : {confidence:.1f}%\n\n"
    title += "\n".join(warnings) + "\n\n"

    if errors:
        title += "🚨 ERREURS DÉTECTÉES :\n\n"
        for error in errors:
            title += f"🔴 {error['name']}\n✅ {error['correction']}\n\n"
    else:
        title += "✅ POSTURE CORRECTE"

    plt.title(title, fontsize=12, loc='left', pad=20)
    plt.axis('off')
    plt.tight_layout()
    plt.show()

# ==========================================
# 🚀 EXÉCUTION
# ==========================================

analyze_yoga_pose()
```

```python
# Conversion du modèle en TensorFlow Lite
import tensorflow as tf

# 🚀 Charger le meilleur modèle entraîné
best_model = tf.keras.models.load_model('best_yoga_model.keras')

# 🔄 Convertir le modèle en format TensorFlow Lite
converter = tf.lite.TFLiteConverter.from_keras_model(best_model)

# ✨ Activer l'optimisation pour réduire la taille du modèle
converter.optimizations = [tf.lite.Optimize.DEFAULT]

# 🔥 Convertir en TensorFlow Lite
tflite_model = converter.convert()

# 📂 Sauvegarde du modèle .tflite
tflite_model_path = "yoga_pose_model.tflite"
with open(tflite_model_path, "wb") as f:
    f.write(tflite_model)

print(f"✅ Modèle TensorFlow Lite exporté sous : {tflite_model_path}")
```

```python
# verification du taille de modele
import os

size_in_kb = os.path.getsize(tflite_model_path) / 1024
print(f"📏 Taille du modèle : {size_in_kb:.2f} KB")
```

```python
# Test du modele

import numpy as np
import tensorflow.lite as tflite

# 🔥 Charger le modèle TFLite
interpreter = tflite.Interpreter(model_path=tflite_model_path)
interpreter.allocate_tensors()

# 📌 Obtenir les détails des entrées/sorties
input_details = interpreter.get_input_details()
output_details = interpreter.get_output_details()

print(f"🎯 Input Shape: {input_details[0]['shape']}")
print(f"🎯 Output Shape: {output_details[0]['shape']}")

# 📸 Générer une image d'exemple aléatoire pour tester
input_data = np.random.rand(1, 224, 224, 3).astype(np.float32)

# 🔮 Faire une prédiction avec le modèle TFLite
interpreter.set_tensor(input_details[0]['index'], input_data)
interpreter.invoke()
output_data = interpreter.get_tensor(output_details[0]['index'])

# 🏆 Affichage du résultat
print(f"🔮 Prédiction du modèle TFLite : Classe {np.argmax(output_data)}")
```
